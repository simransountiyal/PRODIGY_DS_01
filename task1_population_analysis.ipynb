```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/population.csv", skiprows=4)
```

```python
df.head()
```

```python
print(df.columns.tolist())
```

```python
df.head()
```

```python
print(df.shape)
```

```python
year = "2022"
```

```python
population = df[["Country Name", year]].copy()
```

```python
population.head()
```

```python
population = population.dropna()
```

```python
population = population.sort_values(
    by=year,
    ascending=False
)
```

```python
top_10 = population.head(10)
```

```python
top_10
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.bar(
    top_10["Country Name"],
    top_10[year]
)

plt.xlabel("Country")
plt.ylabel("Population")
plt.title("Top 10 Countries by Population in 2022")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

bars = plt.bar(
    top_10["Country Name"],
    top_10[year]
)

plt.xlabel("Country")
plt.ylabel("Population")
plt.title("Top 10 Countries by Population in 2022")

plt.xticks(rotation=45, ha="right")

for bar in bars:
    value = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"{value:,.0f}",
        ha="center",
        va="bottom",
        fontsize=8
    )

plt.tight_layout()
plt.show()
```

## Conclusion

The bar chart shows the population distribution of the 10 most populous countries in 2022. The visualization makes it easy to compare population sizes between countries. The results demonstrate significant differences in population among the selected countries and show how data visualization can make large datasets easier to understand.
